```python
from dotenv import load_dotenv

from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()
llm= ChatGoogleGenerativeAI(model = "gemini-3.7-flash")
```

```python
with open("blood_work.txt", "r") as f:
    blood_report= f.read()

print(blood_report[:200])
```

```python
extraction_prompt= f"""
You are medical data extraction assistant.

From the blood report below. Extract ALL the test values and classify each one as HIGH, LOW and NORMAL 
based on the reference ranges providfed in the report.

Format your reposnse as:
-Test Name: value | Status: HIGH/LOW/NORMAL | Reference: range

Blood_Report:
{blood_report}
"""

extraction_response= llm.invoke(extraction_prompt)
extracted_values= extraction_response.text
print(extracted_values)
```
