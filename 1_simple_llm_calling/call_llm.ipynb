```python
from dotenv import load_dotenv

from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()
```

```python
llm= ChatGoogleGenerativeAI(model = "gemini-3.7-flash")

response = llm.invoke("How many moons does jupiter have?")
print(response.text)
```

```python
llm= ChatGoogleGenerativeAI(model = "gemini-3.7-flash")

response = llm.invoke([
    ["system", "you are helpful assistant and answer in one line"],["human", "How many moons does jupiter have?"]
])
print(response.text)
```

```python
llm= ChatGoogleGenerativeAI(model = "gemini-3.7-flash", temperature=2)

response = llm.invoke([
    ["system", "you are helpful assistant and answer in one line"],["human", "give me a hot gossip tagline for my gossip show"]
])
print(response.text)
```
