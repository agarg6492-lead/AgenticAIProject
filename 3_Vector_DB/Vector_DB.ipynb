```python
import chromadb

client= chromadb.Client()
```

```python
collection= client.create_collection("news_articles")
```

```python
collection.add(
    ids= ["001", "002", "003"],
    documents= ["Apple is leading in smart phone game", "Tesla booked a minor profit of 1 billion", "Apples are high in fiber and vitamin C"]
)
```

```python
#fetch data of collection and embedding
data= collection.get(
    include=["documents", "metadatas", "embeddings"]
)
```

```python
#print data
for i, emb in enumerate(data["embeddings"]):
    print(f"\nDocument {i+1}")
    print("Text:", data["documents"][i])
    print("Embedding Length:", len(emb))
    print("First 10 values:", emb[:10])
```

```python
results= collection.query(
    query_texts= ["This is a query related to elon musk"],
    n_results=2
)
print(results)
```
